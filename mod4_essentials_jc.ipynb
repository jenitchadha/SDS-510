I looked up online and saw Hungry Onion, a public food discussion forum with posts about appliances and also ninja. But I saw there are only 3 posts about it. So instead of explcitly writing ninja food processor, I am going to go with "ninja" only.

```python
driver.get("https://www.hungryonion.org/search?q=ninja")
time.sleep(5)
print(driver.find_element(By.TAG_NAME, "body").text[:2000])
```

```python
#creating a list for urls
post_urls = []
for link in driver.find_elements(By.TAG_NAME, "a"):
    address = link.get_attribute("href")

    if address and address.startswith("https://www.hungryonion.org/t/"):
        if address not in post_urls:
            post_urls.append(address)
for address in post_urls:
    print(address)

print("Total ## of discussion links = ", len(post_urls))
```

```python
# opening one discussion and reading its post.
#I will search for "ninja"
driver.get(post_urls[0])
time.sleep(5)

posts = driver.find_elements(By.CSS_SELECTOR, ".cooked")

for post in posts:
    print(post.text)
```

```python
results = []

for post in posts:
    if "ninja" in post.text.lower():
        results.append({
            "keyword": "ninja",
            "text": post.text,
            "url": driver.current_url
        })
        print(post.text)
```

The above looks pretty cool. Now, I will make a list of keywords and make a loop as I did for the basics assignment.

The list of keywords I'm thinking is: keywords = ["ninja", "blender", "food processor", "air fryer"]

```python
import google_colab_selenium as gs
import time
from selenium.webdriver.common.by import By

driver = gs.Chrome()

keywords = ["ninja", "blender", "food processor", "air fryer", "appliance"]
results = []

for url in post_urls[:5]:
    driver.get(url)
    time.sleep(5)

    posts = driver.find_elements(By.CSS_SELECTOR, ".cooked")

    for post in posts:
        text = post.text

        for word in keywords:
            if word in text.lower():
                results.append({
                    "keyword": word,
                    "text": text,
                    "url": url
                })

    print("Finished:", url)

print("Total matches = ", len(results))
```
